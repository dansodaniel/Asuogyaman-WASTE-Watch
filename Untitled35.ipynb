```python
import os
import folium

os.makedirs("maps", exist_ok=True)

m = folium.Map(
    location=[6.25, 0.05],
    zoom_start=10
)

for _, row in communities.iterrows():
    folium.Marker(
        location=[
            row["Latitude"],
            row["Longitude"]
        ],
        popup=str(row["Community_Name"])
    ).add_to(m)

m.save("maps/asuogyaman_communities.html")

print("Map saved successfully!")
```

```python
records_by_point = (
    collection_records
    .groupby("Point_ID")
    .agg(
        total_kg=("Waste_Collected_kg", "sum"),
        n_records=("Record_ID", "count"),
        missed=("Collection_Status", lambda s: (s == "Missed").sum()),
        overflow=("Collection_Status", lambda s: (s == "Overflow Reported").sum()),
    )
    .reset_index()
)

records_by_point["missed_rate"] = records_by_point["missed"] / records_by_point["n_records"]
records_by_point["overflow_rate"] = records_by_point["overflow"] / records_by_point["n_records"]

hotspots = collection_points.merge(records_by_point, on="Point_ID", how="left").fillna(0)

# weighted hotspot score, 0-100
hotspots["hotspot_score"] = (
    hotspots["missed_rate"] * 45 + hotspots["overflow_rate"] * 55
).clip(0, 100).round(1)

hotspots.sort_values("hotspot_score", ascending=False).head(10)
```

```python
from folium.plugins import HeatMap

def severity_color(score):
    if score >= 35: return "red"
    if score >= 15: return "orange"
    return "green"

def facility_color(row):
    if row["Functional"] == "No": return "red"
    if row["Condition"] == "Poor": return "orange"
    if row["Condition"] == "Fair": return "beige"
    return "green"

m2 = folium.Map(location=[6.25, 0.02], zoom_start=11, tiles="OpenStreetMap")

# --- Layer 1: Collection point hotspots ---
hotspot_layer = folium.FeatureGroup(name="Collection Hotspots", show=True)
for _, row in hotspots.iterrows():
    folium.CircleMarker(
        location=[row["Latitude"], row["Longitude"]],
        radius=6 + (row["hotspot_score"] / 100) * 18,
        color=severity_color(row["hotspot_score"]),
        fill=True, fill_opacity=0.5,
        popup=folium.Popup(
            f"<b>{row['Point_Name']}</b><br>{row['Community_Name']} · {row['Point_Type']}"
            f"<br>Hotspot score: <b>{row['hotspot_score']}</b>"
            f"<br>Missed rate: {row['missed_rate']:.0%} · Overflow rate: {row['overflow_rate']:.0%}"
            f"<br>Total collected: {row['total_kg']:,.0f} kg",
            max_width=250
        )
    ).add_to(hotspot_layer)
hotspot_layer.add_to(m2)

# --- Layer 2: Heat overlay ---
heat_data = [
    [row["Latitude"], row["Longitude"], row["hotspot_score"]]
    for _, row in hotspots.iterrows() if row["hotspot_score"] > 0
]
heat_layer = folium.FeatureGroup(name="Hotspot Heat Overlay", show=False)
HeatMap(heat_data, radius=20, blur=12, max_zoom=10).add_to(heat_layer)
heat_layer.add_to(m2)

# --- Layer 3: Sanitation facilities ---
fac_with_coords = facilities.merge(
    communities[["Community_Name", "Latitude", "Longitude"]],
    on="Community_Name", how="left"
)
facility_layer = folium.FeatureGroup(name="Sanitation Facilities", show=False)
for _, row in fac_with_coords.iterrows():
    folium.CircleMarker(
        location=[row["Latitude"], row["Longitude"]],
        radius=6,
        color=facility_color(row),
        fill=True, fill_opacity=0.7,
        popup=folium.Popup(
            f"<b>{row['Facility_Type']}</b><br>{row['Community_Name']}"
            f"<br>Ownership: {row['Ownership']}<br>Condition: {row['Condition']}"
            f"<br>Functional: {row['Functional']}<br>Est. daily users: {row['Est_Daily_Users']}",
            max_width=250
        )
    ).add_to(facility_layer)
facility_layer.add_to(m2)

folium.LayerControl(collapsed=False).add_to(m2)

os.makedirs("maps", exist_ok=True)
m2.save("maps/asuogyaman_hotspots.html")
m2
```
